####  2.1 Cleaning: Missing values

```python
# Using mode, Because there's only one missing value for "age"
df['age'].fillna(int(df['age'].mode()), inplace=True)
#Replace missing values using the value of the next row, this is only one value in "gender"
# We could use and "Unknown" value, but because it's only one value we don't apply that function
df['gender']=df['gender'].fillna(method='bfill')
#df['age'].unique()
```

```python
#We see the basic statistics values for df
#For the next missing values we'll use median an mode, analysing the statistics and the type of variable. 
df.describe()
```

```python
#Reference: https://www.w3schools.com/python/matplotlib_subplot.asp
#Plot distribution for height
plt.subplot(1, 2, 1)
sns.histplot(df['height'], bins=30, kde=True)
plt.title('Distribution of Height')
plt.xlabel('Height')
plt.ylabel('Frequency')
#Plot distribution for weight
plt.subplot(1, 2, 2)
sns.histplot(df['weight'], bins=30, kde=True)
plt.title('Distribution of Weight')
plt.xlabel('Weight')
plt.ylabel('Frequency')
           
plt.grid()
plt.show()
```

```python
#According to the previous distribution, it is not reccomended apply mean because values are slighlty skewed.
#We apply median
df['height'].fillna(int(df['height'].median()),inplace=True)
df['weight'].fillna(int(df['weight'].median()),inplace=True)
```

```python
a=df['phq1'].unique()
b=df['gad1'].unique()
c=df['epw1'].unique()
print('Unique values for:')
print('phq1:',a,'gad1:',b,'epw1:',c)
```

#### 2.2 Analysing dataset to justify next models

```python
df.describe()
```

According to the statistics results, it will be evaluated the grade anxiety disorder, but first it will be created a new column to classify the total grade of GAD.
Following the next reference: https://www.mdcalc.com/calc/1727/gad7-general-anxiety-disorder7


So in the next steps, it will be applied Interferenctial Statistics

```python
df1=df
```

```python
#Addition of GAD1+...+GAD7
df1["Total_GAD"]=df1['gad1']+df1['gad2']+df1['gad3']+df1['gad4']+df1['gad5']+df1['gad6']+df1['gad7']
df1["Total_GAD"].head()
```

```python
def gdaT(value):
    if value <=9: 
        return "Mild"
    elif value >=10 and value <=14:
        return "Moderate"
    else:
        return "High"
df1["Level_GAD"]=df1["Total_GAD"].apply(gdaT)
df1["Level_GAD"].head()

#data["custom_bins"]=data["MedInc"].apply(custom_discrete)
```

```python
#Plot of Level of GAD
plt.subplot(1, 2, 1)
sns.histplot(df1["Level_GAD"]);

plt.subplot(1, 2, 2)
sns.histplot(df1['grades']);
```

#### 2.3 Inferential Statistics  and Confidence Interval
Interferencial Statistics works with samples, but makes conclusions about the population. Its means that it will calculate values for the population based on the sample. In this case is important to analyse the sample, this analysis is known as Confidence Interval. 
Usually this confidency interval has a value of 95%
