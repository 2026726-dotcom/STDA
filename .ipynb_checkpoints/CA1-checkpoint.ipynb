Normal Range: 1.005–1.030 for adults; 1.001–1.020 for newborns. 
High Values (>1.030): Indicate concentrated urine, often due to dehydration, vomiting, or excess solutes like glucose or protein. 
Low Values (<1.005): Indicate dilute urine, commonly from high fluid intake, diuretic use, or conditions like diabetes insipidus. 
Fixed Value (~1.010): Known as isosthenuria, this suggests the kidneys have lost the ability to concentrate or dilute urine, often signaling kidney dysfunction
