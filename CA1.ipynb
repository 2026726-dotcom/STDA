### Binomial
Binomial Distribution For the binomial distribution, we need to define the number of trials and the probability of success. Let’s say we have 10 trials (e.g., number of agricultors) and we want to see how many of those get an production above a certain threshold.

```python

prod_rice=data['Paddy yield(in Kg)']
# Define threshold for "successful paddy yield"
threshold = prod_rice.mean()  #when it's more than the average
successful_prod = prod_rice > threshold  # Convert to binary outcomes (1 if above threshold)

# Estimate probability of success from data
p_success = np.mean(successful_prod)  # Probability of production above threshold
```

```python
prod_rice.unique()
```

```python
# Define number of trials and success probability
n_trials = 10  # e.g., number of customers in a party
# p_success = 0.5  # Assume a 50% chance of tipping above a threshold

# Generate binomial distribution data
binomial_data = np.random.binomial(n_trials, p_success, 1000)

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(binomial_data, bins=range(n_trials + 2), kde=False, stat='probability')
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successful Prod (Above Threshold)')
plt.ylabel('Probability')
plt.grid()
plt.show()
```

### Poisson
The Poisson Distribution describes the number of events that occur in a fixed interval of time or space when these events happen with a known constant mean rate (( \lambda )) and are independent of the time since the last event.



```python
#Let's supossed the probability for 20 experiments, where we can see how many agricultors can have certain number of hectares
#It means being > than the mean. Use of Poisson.
lambda_rate = data['Hectares '].mean()

# Generate Poisson distribution data
poisson_data = np.random.poisson(lambda_rate, 10) #20 es el numero de experimentos

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(poisson_data, bins=range(0, 10), kde=False, stat='probability')
plt.title('Poisson Distribution (λ = Average Group Size)')
plt.xlabel('No of Hectares')
plt.ylabel('Probability')
plt.grid()
plt.show()
```

```python
lambda_rate = data['Variety'].mean()

# Generate Poisson distribution data
poisson_data = np.random.poisson(lambda_rate, 10) #20 es el numero de experimentos

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(poisson_data, bins=range(0, 8), kde=False, stat='probability')
plt.title('Poisson Distribution (λ = Average Group Size)')
plt.xlabel('No of Hectares')
plt.ylabel('Probability')
plt.grid()
plt.show()
```
