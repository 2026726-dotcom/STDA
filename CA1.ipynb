1. Measures of central tendency 
    - Mean
    - Median
    - Mode

```python
#Mean and Median for NCP and CH2O
meanNCP = data["NCP"].mean()
meanNCP = f"{meanNCP:.2f}"
meanCH2O = data["CH2O"].mean()
meanCH2O = f"{meanCH2O:.2f}"
medianNCP = data["NCP"].median()
medianCH2O = data["CH2O"].median()

print('Mean NCP: ', meanNCP,'Mean CH2O: ', meanCH2O,'\n'
      'Median NCP: ', medianNCP, 'Median CH2O: ', medianCH2O)

```

```python
#conversion float to integer
 
#Mode for Age and TUE
modeNCP = data['NCP'].mode()
modeNCP = int(modeNCP)
modeCH2O = data["CH2O"].mode()
modeCH2O = int(modeCH2O)
print('Mode NCP: ', modeNCP,'Mode CH2O: ', modeCH2O)
```
