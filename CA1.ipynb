```python
#Mean, Median & Mode for Hectares, Nursery area (Cents), Trash(in bundles) and Paddy yield(in Kg) 
#note: 1 cent = 40.47 metros cuadrados (m²) or 100 cents = 1 acre

meanHec = data["Hectares "].mean()
meanHec = f"{meanHec:.2f}"   #to get only 2 decimals 
meanNA = data["Nursery area (Cents)"].mean()
meanNA = f"{meanNA:.2f}"
meanTrash = data["Trash(in bundles)"].mean()
meanTrash = f"{meanTrash:.2f}"
meanPY = data["Paddy yield(in Kg)"].mean()
meanPY = f"{meanPY:.2f}"

medianHec = data["Hectares "].median()
medianNA = data["Nursery area (Cents)"].median()
medianTrash = data["Trash(in bundles)"].median()
medianPY = data["Paddy yield(in Kg)"].median()

modeHec = data["Hectares "].mode()
modeNA = data["Nursery area (Cents)"].mode()
modeTrash = data["Trash(in bundles)"].mode()
modePY = data["Paddy yield(in Kg)"].mode()


print('Hectares > Mean:', meanHec,'Median:', medianHec, 'Mode:',modeHec ,'\n'
      'Nursery area (Cents) > Mean: ', meanNA,'Median: ', medianNA, 'Mode: ',modeNA ,'\n'
      'Trash (in bundles) > Mean: ', meanTrash,'Median: ', medianTrash, 'Mode: ',modeTrash ,'\n'
      'Paddy yield (Kg) > Mean: ', meanPY,'Median: ', medianPY, 'Mode: ',modePY)

```

### Binomial
Binomial Distribution For the binomial distribution, we need to define the number of trials and the probability of success. Let’s say we have 10 trials (e.g., number of agricultors) and we want to see how many of those get an production above a certain threshold.

```python

prod_rice=data['Paddy yield(in Kg)']
# Define threshold for "successful paddy yield"
threshold = prod_rice.mean()  #when it's more than the average
successful_prod = prod_rice > threshold  # Convert to binary outcomes (1 if above threshold)

# Estimate probability of success from data
p_success = np.mean(successful_prod)  # Probability of production above threshold
```

```python
prod_rice.unique()
```

```python
# Define number of trials and success probability
n_trials = 10  # e.g., number of customers in a party
# p_success = 0.5  # Assume a 50% chance of tipping above a threshold

# Generate binomial distribution data
binomial_data = np.random.binomial(n_trials, p_success, 1000)

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(binomial_data, bins=range(n_trials + 2), kde=False, stat='probability')
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of Successful Prod (Above Threshold)')
plt.ylabel('Probability')
plt.grid()
plt.show()
```

### Poisson
The Poisson Distribution describes the number of events that occur in a fixed interval of time or space when these events happen with a known constant mean rate (( \lambda )) and are independent of the time since the last event.



```python
#Let's supossed the probability for 20 experiments, where we can see how many agricultors can have certain number of hectares
#It means being > than the mean. Use of Poisson.
lambda_rate = data['Hectares '].mean()

# Generate Poisson distribution data
poisson_data = np.random.poisson(lambda_rate, 10) #20 es el numero de experimentos

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(poisson_data, bins=range(0, 10), kde=False, stat='probability')
plt.title('Poisson Distribution (λ = Average Group Size)')
plt.xlabel('No of Agricultors')
plt.ylabel('Probability')
plt.grid()
plt.show()
```

```python
#No of Agricultors that can have a specific variety of rice
lambda_rate = data['Variety'].mean()

# Generate Poisson distribution data
poisson_data = np.random.poisson(lambda_rate, 10) #20 es el numero de experimentos

# Plot
plt.figure(figsize=(10, 5))
sns.histplot(poisson_data, bins=range(0, 5), kde=False, stat='probability')
plt.title('Poisson Distribution (λ = Average Group Size)')
plt.xlabel('No of Agricultors')
plt.ylabel('Probability')
plt.grid()
plt.show()
```

### Normal Distribution

Here, we visualize the distribution of Trash(in bundles). We expect to get an normal distribution, because the trash migh  be not very out of the mean. 

```python
#Because the dataset has not normal values, it's still possible to get a normal distribution using quantiles transformation.
#
from sklearn.preprocessing import QuantileTransformer
qt = QuantileTransformer(output_distribution='normal', random_state=42) #this lines use 1000 quantiles per default, a Normal dist
data['Trash(in bundles)_N'] = qt.fit_transform(data[['Trash(in bundles)']])

data.head()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data['Trash(in bundles)_N'], bins=20, kde=True)
plt.title('Normal Distribution of Trash(in bundles) in the production of rice')
plt.xlabel('Trash (in bundles)')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```
